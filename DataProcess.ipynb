```python
import pandas as pd
import numpy as np
import os
```

```python
raw_path = "data/raw/GlobalWeatherRepository.csv"

df = pd.read_csv(raw_path, engine="python")

df.head()
```

```python
df = df.drop_duplicates()
```

```python
df["last_updated"] = pd.to_datetime(
    df["last_updated"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)
```

```python
df["last_updated"].head()
```

```python
df.describe()
```

```python
df["temperature_celsius"].min(), df["temperature_celsius"].max()
```

```python
columns_to_drop = [
    "temperature_fahrenheit",
    "feels_like_fahrenheit",
    "wind_mph",
    "pressure_in",
    "precip_in",
    "visibility_miles",
    "gust_mph"
]

df.drop(columns=columns_to_drop, inplace=True, errors="ignore")
```

```python
from sklearn.preprocessing import MinMaxScaler

numeric_cols = df.select_dtypes(include="number").columns

scaler = MinMaxScaler()

df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

df.head()
```

```python
df["year_month"] = df["last_updated"].dt.to_period("M")
```

```python
monthly_avg = (
    df.groupby("year_month")
    .mean(numeric_only=True)
    .reset_index()
)

monthly_avg.head()
```

```python
os.makedirs("data/processed", exist_ok=True)
df.to_csv(
    "data/processed/global_weather_cleaned_daily.csv",
    index=False
)
```

```python
monthly_avg.to_csv(
    "data/processed/global_weather_cleaned_monthly.csv",
    index=False
)
```
