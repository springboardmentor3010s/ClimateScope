```python
import pandas as pd

df = pd.read_csv(r"D:\global_income_inequality\global_income_inequality.csv")

print("Original Shape:", df.shape)
print(df.head())

df = df.drop_duplicates()

df = df.dropna()

df['Year'] = pd.to_numeric(df['Year'], errors='coerce')

df = df[(df['Year'] >= 2000) & (df['Year'] <= 2023)]

df = df.rename(columns={
    'Gini Index': 'Gini_Index',
    'Average Income (USD)': 'Avg_Income',
    'Top 10% Income Share (%)': 'Top10_Share',
    'Bottom 10% Income Share (%)': 'Bottom10_Share'
})


countries = ['India', 'United States', 'United Kingdom']
df = df[df['Country'].isin(countries)]

df.to_csv(r"D:\cleaned_income_data.csv", index=False)

print("Final Shape:", df.shape)
print("✅ Cleaned file saved at D:\\cleaned_income_data.csv")
```
