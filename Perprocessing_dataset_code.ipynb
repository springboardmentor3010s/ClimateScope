```python
# ================================
# 1. Import Required Libraries
# ================================
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# ================================
# 2. Load Dataset
# ================================
file_path = "/content/GlobalWeatherRepository.csv"
df = pd.read_csv(file_path)

print("Dataset Loaded Successfully!\n")


```

```python
print(df.shape)
df.head()
```

```python
# ================================
# 3. Inspect Dataset Structure
# ================================
print("First 5 Rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nColumn Names:")
print(df.columns)

```

```python
# ================================
# 4. Identify Missing Values
# ================================
print("\nMissing Values Count:")
print(df.isnull().sum())

print("\nMissing Values Percentage:")
print((df.isnull().sum() / len(df)) * 100)

df = df.drop_duplicates()

```

```python
# ================================
# 5. Remove Unwanted Columns
# (Modify based on your dataset)
# ================================
unwanted_columns = [
    'temperature_fahrenheit',
    'wind_mph',
    'pressure_in',
    'precip_in',
    'feels_like_fahrenheit',
    'visibility_miles',
    'gust_mph',
    'last_updated_epoch',
    'wind_degree',
    'timezone',
    'moonrise',
    'moonset',
    'moon_phase',
    'moon_illumination',
    'air_quality_Carbon_Monoxide',
    'air_quality_Ozone',
    'air_quality_Nitrogen_dioxide',
    'air_quality_Sulphur_dioxide',
    'air_quality_PM2.5',
    'air_quality_PM10',
    'air_quality_gb-defra-index'
]
df.drop(columns=[col for col in unwanted_columns if col in df.columns], inplace=True)

print("\nColumns After Removal:")
print(df.columns)

```

```python
print(df.shape)
df.head()
```

```python
# ================================
# 6. Handle Missing Values
# ================================

# Fill numerical columns with median
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

# Fill categorical columns with mode
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("\nMissing Values After Handling:")
print(df.isnull().sum())

df['last_updated'] = pd.to_datetime(df['last_updated'])

# Keep only latest 6 months
df = df[df['last_updated'] > df['last_updated'].max() - pd.DateOffset(months=6)]


```

```python
# ================================
# 7. Detect & Handle Outliers (Using IQR Method)
# ================================
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = np.where(df[col] < lower, lower,
              np.where(df[col] > upper, upper, df[col]))

print("\nOutliers Handled Using IQR Method")

# ================================
# 8. Convert Units (Example)
# ================================
# Convert Temperature from Fahrenheit to Celsius (if needed)
if 'temperature_f' in df.columns:
    df['temperature_c'] = (df['temperature_f'] - 32) * 5/9

# Convert Wind Speed km/h to m/s (if needed)
if 'wind_speed_kmh' in df.columns:
    df['wind_speed_ms'] = df['wind_speed_kmh'] / 3.6

print("\nUnit Conversion Completed (if applicable)")

```

```python
# ================================
# 9. Normalize Numerical Values
# ================================
scaler = MinMaxScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])

print("\nNormalization Completed")

```

```python
# ================================
# 10. Convert Date Column & Aggregate
# ================================
# Convert date column (modify column name if needed)
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'])
    df['year_month'] = df['date'].dt.to_period('M')

    monthly_avg = df.groupby('year_month')[num_cols].mean()

    print("\nMonthly Aggregated Data:")
    print(monthly_avg.head())

```

```python
# ================================
# 11. Save Cleaned Dataset
# ================================
df.to_csv("GlobalWeatherRepository_Cleaned.csv", index=False)
print("\nCleaned Dataset Saved Successfully!")
```
