```python
df.drop_duplicates(inplace=True)
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(" ", "_")
```

```python
df["temperature_celsius"] = pd.to_numeric(df["temperature_celsius"])
```

```python
df.drop('temperature_fahrenheit', axis=1, inplace=True)
```

```python
df['last_updated'] = pd.to_datetime(df['last_updated'])
df['last_updated'].head()
```

```python
missing_counts = df.isnull().sum()
missing_percent = (missing_counts / len(df)) * 100
missing_summary = pd.DataFrame({'missing_count': missing_counts, 'missing_percent': missing_percent})
missing_summary.sort_values(by='missing_percent', ascending=False)
```

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Import Libraries & Load Cleaned CSV
```

```python
df = pd.read_csv("cleaned_weather.csv.gz")

df.head()
```

```python
df = pd.read_csv("cleaned_weather.csv")

# Convert time column
df['last_updated'] = pd.to_datetime(df['last_updated'])

# Create date column for aggregation
df['date'] = df['last_updated'].dt.date
```

```python
#Statistical Analysis (Summary Statistics)
```

```python
df.describe()
```

```python
df.info()
```

```python
#Distribution Analysis (Understand Data Spread)
```

```python
plt.figure()
sns.histplot(df['temperature_celsius'], kde=True)
plt.title("Temperature Distribution")
plt.show()
```

```python
plt.figure()
sns.histplot(df['precip_mm'], kde=True)
plt.title("Precipitation Distribution (mm)")
plt.show()
```

```python
plt.figure()
sns.histplot(df['humidity'], kde=True)
plt.title("Humidity Distribution")
plt.show()
```

```python
extreme_humidity = df[df['humidity'] > df['humidity'].quantile(0.95)]
extreme_humidity.shape
```

```python
#Correlation Analysis
```

```python
corr = df.corr(numeric_only=True)
corr
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(corr, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

#DAILY TREND ANALYSIS

```python
daily_df = df.groupby('date').mean(numeric_only=True)
```

```python
#DAILY TEMPERATURE TREND
```

```python
plt.figure(figsize=(10,5))
plt.plot(daily_df.index, daily_df['temperature_celsius'])
plt.title("Daily Average Temperature Trend")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
#Seasonal Analysis (Monthly)
```

```python
df['month'] = df['last_updated'].dt.month

monthly_temp = df.groupby('month')['temperature_celsius'].mean()
```

```python
monthly_temp.plot(marker='o')
plt.title("Average Monthly Temperature")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
#Extreme Weather Events
```

```python
extreme_heat = df[df['temperature_celsius'] > df['temperature_celsius'].quantile(0.95)]
len(extreme_heat)
```

```python
extreme_rain = df[df['precip_mm'] > df['precip_mm'].quantile(0.95)]
len(extreme_rain)
```

```python
#Regional Comparison
```

```python
country_temp = df.groupby('country')['temperature_celsius'].mean().sort_values(ascending=False)
```

```python
country_temp.head(10).plot(kind='bar')
plt.title("Average Temperature by Country")
plt.ylabel("Temperature (°C)")
plt.show()
```
