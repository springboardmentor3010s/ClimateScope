```python
import pandas as pd
```

```python
df = pd.read_csv("GlobalWeatherRepository.csv", low_memory=False)
```

```python
df.head()
df.info()
df.shape
```

```python
df.isnull().sum()
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df['last_updated'] = pd.to_datetime(df['last_updated'], dayfirst=True)
```

```python
df['year'] = df['last_updated'].dt.year
df['month'] = df['last_updated'].dt.month
df['day'] = df['last_updated'].dt.day
```

```python
df['condition_text'] = df['condition_text'].str.lower()
df['country'] = df['country'].str.title()
```

```python
df.describe()
```

```python
df.to_csv("GlobalWeatherRepository_CLEANED.csv", index=False)
print("File saved")
```
