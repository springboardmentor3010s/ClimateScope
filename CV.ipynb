```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("GlobalWeatherRepository_CLEANED.csv")
```

```python
df[['temperature_celsius', 'humidity', 'wind_kph', 'pressure_mb']].describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['temperature_celsius'], kde=True)
plt.title("Temperature Distribution")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['temperature_celsius'])
plt.title("Temperature Boxplot")
plt.xlabel("Temperature (°C)")
plt.show()
```

```python
df[['temperature_celsius','humidity','wind_kph','pressure_mb']].corr()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(
    df[['temperature_celsius','humidity','wind_kph','pressure_mb']].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Between Weather Variables")
plt.show()
```

```python
monthly_stats = df.groupby('month')['temperature_celsius'].agg(['mean', 'min', 'max', 'std'])
monthly_stats
```

```python
plt.figure(figsize=(8,5))
monthly_stats['mean'].plot(marker='o')
plt.title("Monthly Average Temperature")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
yearly_stats = df.groupby('year')['temperature_celsius'].agg(['mean', 'min', 'max', 'std'])
yearly_stats
```

```python
plt.figure(figsize=(8,5))
yearly_stats['mean'].plot(marker='o')
plt.title("Yearly Average Temperature Trend")
plt.xlabel("Year")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
extreme_high = df[df['temperature_celsius'] >= high_threshold]
extreme_low = df[df['temperature_celsius'] <= low_threshold]
```

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
df = pd.read_csv("GlobalWeatherRepository_CLEANED.csv")
```

```python
# Monthly Average Temperature
monthly_avg = df.groupby('month')['temperature_celsius'].mean().reset_index()

# Yearly Average Temperature
yearly_avg = df.groupby('year')['temperature_celsius'].mean().reset_index()

# Country Average Temperature
country_avg = df.groupby('country')['temperature_celsius'].mean().reset_index()

# Correlation Matrix
corr = df[['temperature_celsius','humidity','wind_kph','pressure_mb']].corr()

# Top 10 Hottest Countries
top_countries = country_avg.sort_values(
    by='temperature_celsius',
    ascending=False
).head(10)
```

```python
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        "Monthly Temperature Trend",
        "Temperature vs Humidity",
        "Top 10 Hottest Countries",
        "Correlation Heatmap"
    )
)
```

```python
fig.add_trace(
    go.Scatter(
        x=monthly_avg['month'],
        y=monthly_avg['temperature_celsius'],
        mode='lines+markers',
        name='Monthly Avg Temp'
    ),
    row=1, col=1
)
```

```python
fig.add_trace(
    go.Scatter(
        x=df['temperature_celsius'],
        y=df['humidity'],
        mode='markers',
        marker=dict(size=4),
        name='Temp vs Humidity'
    ),
    row=1, col=2
)
```

```python
fig.add_trace(
    go.Bar(
        x=top_countries['country'],
        y=top_countries['temperature_celsius'],
        name='Top Countries'
    ),
    row=2, col=1
)
```

```python
fig.add_trace(
    go.Heatmap(
        z=corr.values,
        x=corr.columns,
        y=corr.columns,
        colorscale='Viridis'
    ),
    row=2, col=2
)
```

```python
fig.update_layout(
    height=800,
    width=1100,
    title_text="Climate Scope - Weather Analysis Dashboard",
    showlegend=False
)

fig.show()
```
