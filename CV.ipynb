```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("GlobalWeatherRepository_CLEANED.csv")
```

```python
df[['temperature_celsius', 'humidity', 'wind_kph', 'pressure_mb']].describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['temperature_celsius'], kde=True)
plt.title("Temperature Distribution")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['temperature_celsius'])
plt.title("Temperature Boxplot")
plt.xlabel("Temperature (°C)")
plt.show()
```

```python
df[['temperature_celsius','humidity','wind_kph','pressure_mb']].corr()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(
    df[['temperature_celsius','humidity','wind_kph','pressure_mb']].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Between Weather Variables")
plt.show()
```

```python
monthly_stats = df.groupby('month')['temperature_celsius'].agg(['mean', 'min', 'max', 'std'])
monthly_stats
```

```python
plt.figure(figsize=(8,5))
monthly_stats['mean'].plot(marker='o')
plt.title("Monthly Average Temperature")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
yearly_stats = df.groupby('year')['temperature_celsius'].agg(['mean', 'min', 'max', 'std'])
yearly_stats
```

```python
plt.figure(figsize=(8,5))
yearly_stats['mean'].plot(marker='o')
plt.title("Yearly Average Temperature Trend")
plt.xlabel("Year")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
high_threshold = df['temperature_celsius'].quantile(0.95)
low_threshold = df['temperature_celsius'].quantile(0.05)
```

```python
extreme_high = df[df['temperature_celsius'] >= high_threshold]
extreme_low = df[df['temperature_celsius'] <= low_threshold]
```

```python
extreme_high = df[df['temperature_celsius'] >= high_threshold]
extreme_low = df[df['temperature_celsius'] <= low_threshold]
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.scatter(extreme_high['month'], extreme_high['temperature_celsius'])
plt.title("Extreme High Temperature Events")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
country_avg = df.groupby('country')['temperature_celsius'].mean().sort_values(ascending=False)
country_avg.head(10)
```

```python
country_avg.head(10).plot(kind='bar', figsize=(10,5))
plt.title("Top 10 Hottest Countries (Average Temperature)")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
humidity_avg = df.groupby('country')['humidity'].mean().sort_values(ascending=False)
humidity_avg.head(10)
```

```python
#Scatterplot
plt.figure(figsize=(6,5))
plt.scatter(df['temperature_celsius'], df['humidity'], alpha=0.3)
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity")
plt.title("Temperature vs Humidity")
plt.show()
```

```python
#Heatmap
import seaborn as sns
plt.figure(figsize=(8,6))
sns.heatmap(
    df[['temperature_celsius','humidity','wind_kph','pressure_mb']].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
#Line Chart
monthly_avg = df.groupby('month')['temperature_celsius'].mean()
monthly_avg.plot(marker='o')
plt.title("Monthly Average Temperature Trend")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
#Bar Chart
country_avg = df.groupby('country')['temperature_celsius'].mean().head(10)
country_avg.plot(kind='bar')
plt.title("Top 10 Countries by Avg Temperature")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
#Choropleth Maps
import plotly.express as px
country_avg = df.groupby('country')['temperature_celsius'].mean().reset_index()
fig = px.choropleth(
    country_avg,
    locations="country",
    locationmode="country names",
    color="temperature_celsius",
    title="Global Average Temperature Distribution"
)
fig.show()
```
