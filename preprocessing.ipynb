```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("GlobalWeatherRepository.xlsx")
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(" ", "_")
```

```python
df["temperature_celsius"] = pd.to_numeric(df["temperature_celsius"])
```

```python
df['last_updated'] = pd.to_datetime(df['last_updated'])
df['last_updated'].head()
```

```python
missing_counts = df.isnull().sum()
missing_percent = (missing_counts / len(df)) * 100
missing_summary = pd.DataFrame({'missing_count': missing_counts, 'missing_percent': missing_percent})
missing_summary.sort_values(by='missing_percent', ascending=False)
```

```python
df.to_csv("cleaned_weather.csv.gz", index=False)
```
