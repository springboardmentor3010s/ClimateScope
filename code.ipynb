```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\GEETHIKA\Downloads\archive (6)\GlobalWeatherRepository.csv")
```

```python
df.head()
df.shape
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df.fillna({
    'temperature_celsius': df['temperature_celsius'].mean(),
    'humidity': df['humidity'].median(),
    'wind_kph': df['wind_kph'].mean(),
    'pressure_mb': df['pressure_mb'].mean()
}, inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['last_updated'] = pd.to_datetime(df['last_updated'])
```

```python
df.dtypes
```

```python
df = df[df['humidity'] <= 100]
df = df[df['humidity'] >= 0]

df = df[df['wind_kph'] >= 0]

df = df[(df['pressure_mb'] > 850) & (df['pressure_mb'] < 1100)]
```

```python
df.drop(columns=[
    'temperature_fahrenheit',
    'feels_like_fahrenheit',
    'wind_mph',
    'pressure_in',
    'precip_in',
    'visibility_miles'
], inplace=True)
```

```python
df['year'] = df['last_updated'].dt.year
df['month'] = df['last_updated'].dt.month
df['day'] = df['last_updated'].dt.day
df['hour'] = df['last_updated'].dt.hour
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

numeric_cols = [
    'temperature_celsius',
    'humidity',
    'wind_kph',
    'pressure_mb',
    'precip_mm',
    'visibility_km',
    'uv_index'
]

df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
```

```python
monthly_weather = (
    df.set_index('last_updated')
      .select_dtypes(include='number')
      .resample('M')
      .mean()
)
```

```python
df.to_csv("global_weather_cleaned.csv", index=False)
monthly_weather.to_csv("global_weather_monthly.csv")
```
