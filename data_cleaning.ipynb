```python
cs['calculate_kph'] = (cs['wind_kph'] * 0.621371)
cs[['wind_kph', 'wind_mph', 'calculate_kph']].head()
cs['calculate_visibility_km'] = (cs['visibility_miles'] * 1.60934).round()
cs[['calculate_visibility_km', 'visibility_km', 'visibility_miles']].head()
cs['calculated_fahrenheit'] = (cs['temperature_celsius'] * 9/5) + 32
cs[['calculated_fahrenheit', 'temperature_celsius']].head()
cs.drop(columns='temperature_fahrenheit', inplace= True)
```

## deletiing the redandent data 

```python
drop_columns_list = [
    'feels_like_fahrenheit', 
    'wind_mph', 
    'gust_mph', 
    'visibility_miles', 
    'pressure_in', 
    'precip_in'
]

cs.drop(columns = drop_columns_list, inplace= True)
```

```python
cs.drop(columns = ['calculate_kph','calculate_visibility_km','calculated_fahrenheit'], inplace= True)
```

### Rename the column name

```python
new_columns_name = []
for i in cs.columns:
    new_columns_name.append((i.upper()))
```

```python
cs.columns = new_columns_name
cs.head()
```

## Indivisual Column work with min and max value


```python
air_quality_columns = ['AIR_QUALITY_CARBON_MONOXIDE', 'AIR_QUALITY_SULPHUR_DIOXIDE', 'AIR_QUALITY_PM10' ]
for i in air_quality_columns:
    cs.loc[cs[i] < 0, i] = np.nan
print(cs[air_quality_columns].isnull().sum())

## fill the value with median with
for i in air_quality_columns:
    median_value = cs[i].median()
    cs[i].fillna(median_value, inplace =True)
```

```python
## remove the Outliers

outiler_list = ['WIND_KPH', 'GUST_KPH']

for i in outiler_list:
    cs.loc[cs[i] > 750, i] = np.nan

    cs[i].fillna(cs[i].median(), inplace= True)

### remove outliers from the pressure MB

cs.loc[cs['PRESSURE_MB'] > 1300, 'PRESSURE_MB'] = np.nan
cs['PRESSURE_MB'].fillna(cs['PRESSURE_MB'].median(), inplace= True)


### removing the outliers from Air Qulity PM

cs.loc[cs['AIR_QUALITY_PM10'] > 1600 , 'AIR_QUALITY_PM10'] = np.nan
cs['AIR_QUALITY_PM10'].fillna(cs['AIR_QUALITY_PM10'].median(), inplace= True)

### removing the outliers from Air_quality_carbon_monoxide

cs.loc[cs['AIR_QUALITY_CARBON_MONOXIDE'] > 1600 , 'AIR_QUALITY_CARBON_MONOXIDE'] = np.nan
cs['AIR_QUALITY_CARBON_MONOXIDE'].fillna(cs['AIR_QUALITY_CARBON_MONOXIDE'].median(), inplace= True)

```

## chaning in the data type 

```python
cs['last_updated'] = pd.to_datetime(cs['last_updated'])
```
