```python
df['temperature_celsius'] = df['temperature_celsius'].fillna(df['temperature_celsius'].mean())

df['humidity'] = df['humidity'].fillna(df['humidity'].median())

df['precip_mm'] = df['precip_mm'].fillna(0)

df['wind_kph'] = df['wind_kph'].fillna(df['wind_kph'].mean())

df['pressure_mb'] = df['pressure_mb'].fillna(df['pressure_mb'].mean())
```

```python
df = df.drop_duplicates()
```

```python
df = df[(df['humidity'] >= 0) & (df['humidity'] <= 100)]
df = df[df['precip_mm'] >= 0]
df = df[df['wind_kph'] >= 0]
```

```python
df['last_updated'] = pd.to_datetime(df['last_updated'], errors='coerce')
```

```python
df['last_updated'].dtype
```

```python
df['month'] = df['last_updated'].dt.month
df['day'] = df['last_updated'].dt.day
df['hour'] = df['last_updated'].dt.hour
df['year'] = df['last_updated'].dt.year
```

```python
monthly_df = df.groupby(['country', 'year', 'month']).agg({
    'temperature_celsius': 'mean',
    'wind_kph': 'mean',
    'precip_mm': 'sum',
    'humidity': 'mean',
    'pressure_mb': 'mean',
    'uv_index': 'mean'
}).reset_index()

monthly_df.head()
```

```python
import os
os.makedirs("data", exist_ok=True)
monthly_df.to_csv("data/cleaned_weather_monthly.csv", index=False)
```

```python
import pycountry

def get_iso3(country_name):
    try:
        return pycountry.countries.lookup(country_name).alpha_3
    except:
        return None

monthly_df['iso3'] = monthly_df['country'].apply(get_iso3)
```

```python
import plotly.express as px

fig = px.choropleth(
    monthly_df,
    locations='iso3',
    locationmode='ISO-3',
    color='temperature_celsius',
    title='Global Temperature Distribution'
)

fig.show()
```
