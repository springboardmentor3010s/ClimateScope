```python
import pandas as pd
import numpy as np
```

```python
file_path = r"data\processed\global_weather_cleaned_daily.csv"

df = pd.read_csv(
    file_path,
    parse_dates=["last_updated"]
)

print("Dataset Shape:", df.shape)
df.head()
```

```python
df.dtypes
```

```python
df = df[df['wind_kph'] < 200]
df = df[(df['pressure_mb'] > 900) & (df['pressure_mb'] < 1100)]
df.replace(-9999, np.nan, inplace=True)
print("Final Dataset Shape:", df.shape)
```

```python
df.describe()
```

```python
weather_cols = [
    'temperature_celsius',
    'humidity',
    'wind_kph',
    'precip_mm',
    'pressure_mb',
    'uv_index'
]

df[weather_cols].describe()
```

```python
df[weather_cols].quantile([0.25, 0.5, 0.75, 0.95, 0.99])
```

```python
corr_matrix = df[weather_cols].corr()

corr_matrix
```

```python
df['year'] = df['last_updated'].dt.year
df['month'] = df['last_updated'].dt.month

monthly_avg = df.groupby(['year', 'month']).agg({
    'temperature_celsius': 'mean',
    'precip_mm': 'mean',
    'wind_kph': 'mean',
    'humidity': 'mean'
}).reset_index()

monthly_avg.head()
```

```python
monthly_avg.to_csv("data/processed/monthly_weather_summary.csv", index=False)
```

```python
yearly_avg = df.groupby('year').agg({
    'temperature_celsius': 'mean',
    'precip_mm': 'mean',
    'wind_kph': 'mean'
}).reset_index()

yearly_avg
```

```python
yearly_avg.to_csv("data/processed/yearly_weather_summary.csv", index=False)
```

```python
country_avg = df.groupby('country').agg({
    'temperature_celsius': 'mean',
    'precip_mm': 'mean',
    'wind_kph': 'mean',
    'humidity': 'mean'
}).reset_index()

country_avg.head()
```

```python
country_avg.to_csv("data/processed/country_weather_summary.csv", index=False)
```

```python

top10_hot = country_avg.sort_values(
    by='temperature_celsius',
    ascending=False
).head(10)

top10_hot
```

```python

top10_hot.to_csv("data/processed/top10_hottest_countries.csv", index=False)
```

```python
top10_rain = country_avg.sort_values(
    by='precip_mm',
    ascending=False
).head(10)

top10_rain
top10_rain.to_csv("data/processed/top10_rainiest_countries.csv", index=False)
```

```python
temp_threshold = df['temperature_celsius'].quantile(0.95)
rain_threshold = df['precip_mm'].quantile(0.95)
wind_threshold = df['wind_kph'].quantile(0.95)
```

```python

extreme_heat = df[df['temperature_celsius'] > temp_threshold]
extreme_rain = df[df['precip_mm'] > rain_threshold]
extreme_wind = df[df['wind_kph'] > wind_threshold]
```

```python
print("Extreme Heat Events:", extreme_heat.shape[0])
print("Extreme Rain Events:", extreme_rain.shape[0])
print("Extreme Wind Events:", extreme_wind.shape[0])
```

```python
extreme_heat.to_csv("data/processed/extreme_heat_events.csv", index=False)
extreme_rain.to_csv("data/processed/extreme_rain_events.csv", index=False)
extreme_wind.to_csv("data/processed/extreme_wind_events.csv", index=False)
```

```python
global_summary = pd.DataFrame({
    "Metric": [
        "Average Temperature",
        "Average Rainfall",
        "Average Wind Speed",
        "Average Humidity"
    ],
    "Value": [
        df['temperature_celsius'].mean(),
        df['precip_mm'].mean(),
        df['wind_kph'].mean(),
        df['humidity'].mean()
    ]
})

global_summary
global_summary.to_csv("data/processed/global_kpi_summary.csv", index=False)
```
