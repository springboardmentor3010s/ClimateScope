```python
import pandas as pd
import numpy as np
```

```python
file_path = r"data\processed\global_weather_cleaned_daily.csv"

df = pd.read_csv(
    file_path,
    parse_dates=["last_updated"]
)

print("Dataset Shape:", df.shape)
df.head()
```

```python
df.dtypes
```

```python
df = df[df['wind_kph'] < 200]
df = df[(df['pressure_mb'] > 900) & (df['pressure_mb'] < 1100)]
df.replace(-9999, np.nan, inplace=True)
print("Final Dataset Shape:", df.shape)
```

```python
df.describe()
```

```python
weather_cols = [
    'temperature_celsius',
    'humidity',
    'wind_kph',
    'precip_mm',
    'pressure_mb',
    'uv_index'
]
df[weather_cols].describe()
```

```python
df[weather_cols].quantile([0.25, 0.5, 0.75, 0.95, 0.99])
```

```python
corr_matrix = df[weather_cols].corr()
corr_matrix
```

```python
df['year'] = df['last_updated'].dt.year
df['month'] = df['last_updated'].dt.month

monthly_avg = df.groupby(['year', 'month']).agg({
    'temperature_celsius': 'mean',
    'precip_mm': 'mean',
    'wind_kph': 'mean',
    'humidity': 'mean'
}).reset_index()

monthly_avg
```

```python
yearly_avg = df.groupby('year').agg({
    'temperature_celsius': 'mean',
    'precip_mm': 'mean',
    'wind_kph': 'mean'
}).reset_index()

yearly_avg
```

```python
country_avg = df.groupby('country').agg({
    'temperature_celsius': 'mean',
    'precip_mm': 'mean',
    'wind_kph': 'mean',
    'humidity': 'mean'
}).reset_index()

country_avg.head()
```

```python

top10_hot = country_avg.sort_values(
    by='temperature_celsius',
    ascending=False
).head(10)

top10_hot
```

```python
top10_rain = country_avg.sort_values(
    by='precip_mm',
    ascending=False
).head(10)

top10_rain
```

```python
temp_threshold = df['temperature_celsius'].quantile(0.95)
rain_threshold = df['precip_mm'].quantile(0.95)
wind_threshold = df['wind_kph'].quantile(0.95)
```

```python

extreme_heat = df[df['temperature_celsius'] > temp_threshold]
extreme_rain = df[df['precip_mm'] > rain_threshold]
extreme_wind = df[df['wind_kph'] > wind_threshold]
```

```python
print("Extreme Heat Events:", extreme_heat.shape[0])
print("Extreme Rain Events:", extreme_rain.shape[0])
print("Extreme Wind Events:", extreme_wind.shape[0])
```

```python
global_summary = pd.DataFrame({
    "Metric": [
        "Average Temperature",
        "Average Rainfall",
        "Average Wind Speed",
        "Average Humidity"
    ],
    "Value": [
        df['temperature_celsius'].mean(),
        df['precip_mm'].mean(),
        df['wind_kph'].mean(),
        df['humidity'].mean()
    ]
})

global_summary
```

1️⃣ Statistical Analysis of Weather Variables

To understand the overall behavior of global weather patterns, descriptive statistical analysis was performed on key meteorological variables including temperature, humidity, precipitation, wind speed, pressure, and UV index.

🔹 Distribution Analysis

The dataset consists of over 120,000 weather observations across multiple global locations. Statistical summaries revealed:

Temperature values range from extreme cold conditions (below freezing) to very high temperatures typical of tropical and desert climates.

Precipitation distribution is highly right-skewed, indicating that most days experience little to no rainfall, while a small percentage of days experience heavy rainfall events.

Wind speed values are mostly concentrated in moderate ranges, with few extreme wind events.

Humidity levels show broad variation depending on geographic location and climate type.

This distribution analysis helps establish the variability and global diversity of climate conditions.

2️⃣ Correlation Analysis

A correlation matrix was computed to examine relationships between major weather variables.

🔹 Key Observations:

Humidity and Precipitation show a positive correlation, indicating that higher humidity levels are often associated with rainfall events.

Temperature and Humidity demonstrate mild inverse relationships in certain regions, reflecting dry heat conditions in arid climates.

Wind speed and precipitation show weak to moderate relationships, suggesting wind contributes to storm systems but is not always directly proportional to rainfall.

Pressure tends to vary inversely during extreme weather events, particularly during storms.

Correlation analysis helps identify interactions between atmospheric variables and supports climate behavior understanding.

3️⃣ Seasonal Patterns & Time-Series Trends

To analyze climate behavior over time, monthly and yearly aggregations were performed.

🔹 Monthly Trends

Monthly averages revealed clear seasonal cycles:

Temperature peaks during mid-year months in many regions, indicating summer season effects.

Cooler temperatures are observed during early and late months of the year.

Rainfall patterns vary significantly by region, with some countries experiencing monsoon-driven seasonal spikes.

Wind speed shows mild seasonal fluctuation but remains relatively stable compared to precipitation.

These seasonal variations confirm predictable climate cycles and demonstrate temporal trends in global weather patterns.

4️⃣ Regional Comparison Across Countries

Country-wise aggregation was performed to compare average weather conditions across regions.

🔹 Regional Insights:

Tropical and desert regions show consistently higher average temperatures.

Coastal and equatorial countries display higher humidity levels.

Certain regions experience significantly higher rainfall due to monsoon systems or tropical storm activity.

Wind intensity varies across geographical terrain, with some regions showing elevated wind averages.

Top 10 hottest and rainiest countries were identified to highlight climatic extremes across regions.

This comparative analysis helps visualize how climate differs globally.

5️⃣ Identification of Extreme Weather Events

Extreme weather events were identified using the 95th percentile threshold for temperature, precipitation, and wind speed.

🔹 Extreme Heat Events:

Days with temperatures above the 95th percentile were classified as extreme heat days. These are concentrated primarily in warmer climate zones.

🔹 Extreme Rainfall Events:

Precipitation values above the 95th percentile indicate heavy rainfall events. These are generally localized but intense.

🔹 Extreme Wind Events:

Wind speeds above the upper threshold represent potential storm or severe weather conditions.

This percentile-based approach ensures objective and statistically grounded identification of anomalies.
