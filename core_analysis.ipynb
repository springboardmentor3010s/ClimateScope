## THE SEASON FIND OUT 

```python
df['LAST_UPDATED'] = pd.to_datetime(df['LAST_UPDATED'])
month =(df['LAST_UPDATED']).dt.month
df['MONTH'] = month
print(df['MONTH'])
def find_season(month_num):
    if month_num in [12,1,2]:
        return 'Winter'
    elif month_num in [3,4,5]:
        return 'Summar'
    elif month_num in [6,7,8,9]:
        return 'Raining'
    elif month_num in [10,11]:
        return 'Post_mansoon'
    else:
        return 'Unknown'
df['SEASON'] = df['MONTH'].apply(find_season)
print(df['SEASON'].unique())
```

# THE YEAR FINDING


```python
year= df['LAST_UPDATED'].dt.year
df['YEAR'] = year
```

# THE TREAD OF SEASON AND WEATHER CONDTION 

```python
monthly_pattern = df.groupby('MONTH')['TEMPERATURE_CELSIUS'].mean().reset_index()
plt.figure(figsize=(10, 5))
plt.plot(monthly_pattern['MONTH'], monthly_pattern['TEMPERATURE_CELSIUS'], 
         marker='o', color='red', linewidth=2, markersize=8)

# Graph ko thoda sajate hain
plt.title(' Monthly Average Temperature', fontsize=14)
plt.xlabel('Month (1 = Jan, 12 = Dec)', fontsize=12)
plt.ylabel('Average Temperature (°C)', fontsize=12) 
plt.grid(True)

# Graph show karna
plt.show()
```

```python
year_pettern = df.groupby('YEAR')['TEMPERATURE_CELSIUS'].mean().reset_index()
print(year_pettern)
year_pettern['YEAR'] = year_pettern['YEAR'].astype(str)
plt.figure(figsize=(8,5))
plt.plot(year_pettern['YEAR'], year_pettern['TEMPERATURE_CELSIUS'], color='blue', marker='o', linewidth=2, markersize=8)
plt.title("Year Tread of Temperature")
plt.xlabel("Average Temperature", fontsize=12)
plt.ylabel("The Number Of Years", fontsize=12)
plt.grid(True, axis='y', linestyle='-', alpha=1)
plt.show()

```

# EXtrame Weather Events

```python
hottest_places = df.nlargest(5, 'TEMPERATURE_CELSIUS')
print(hottest_places[['COUNTRY', 'LOCATION_NAME', 'TEMPERATURE_CELSIUS', 'CONDITION_TEXT']])

coldest_places = df.nsmallest(5, 'TEMPERATURE_CELSIUS')
print(coldest_places[['COUNTRY', 'LOCATION_NAME', 'TEMPERATURE_CELSIUS', 'CONDITION_TEXT']])
```

```python
top_5_country = df.groupby('COUNTRY')['TEMPERATURE_CELSIUS'].max().nlargest(5).reset_index()
plt.figure(figsize=(7,5))
sns.barplot(data=top_5_country, x='COUNTRY', hue='COUNTRY', y='TEMPERATURE_CELSIUS', palette='rocket', legend=False)
plt.title("Top 5 Hottest Unique Country", fontsize=14)
plt.xlabel("COUNTRY / Location", fontsize=12)
plt.ylabel("Maximum Temperature (°C)", fontsize=12)
plt.xticks(rotation=45) 

plt.tight_layout()
plt.show()
```

```python
top_5_cities = df.groupby('LOCATION_NAME')['TEMPERATURE_CELSIUS'].max().nlargest(5).reset_index()
plt.Figure(figsize=(9,5))
sns.barplot(data=top_5_cities, x='LOCATION_NAME', hue='LOCATION_NAME', y='TEMPERATURE_CELSIUS', palette='Reds_r', legend=False)
plt.title("Top 5 Hottest Unique Cities", fontsize=14)
plt.xlabel("City / Location", fontsize=12)
plt.ylabel("Maximum Temperature (°C)", fontsize=12)
plt.xticks(rotation=45) 
plt.tight_layout()
plt.show()
```

```python
low_5_cities =df.groupby('LOCATION_NAME')['TEMPERATURE_CELSIUS'].min().nsmallest().reset_index()
print(low_5_cities)
plt.figure(figsize=(7,5))
sns.barplot(data=low_5_cities, x='LOCATION_NAME', hue='LOCATION_NAME', y='TEMPERATURE_CELSIUS', palette='Blues_r', legend=False)
plt.title("Top 5 Coldest Unique Cities", fontsize=14)
plt.xlabel("City / Location", fontsize=12,)
plt.ylabel("Minimum Temperature (°C)", fontsize=12)
plt.xticks(rotation=45) 
plt.tight_layout()
plt.show()
```

```python
low_5_COUNTRY =df.groupby('COUNTRY')['TEMPERATURE_CELSIUS'].min().nsmallest().reset_index()
print(low_5_COUNTRY)
plt.figure(figsize=(7,5))
sns.barplot(data=low_5_COUNTRY, x='COUNTRY', hue='COUNTRY', y='TEMPERATURE_CELSIUS', palette='RdBu', legend=False)
plt.title("Top 5 Coldest Unique Country", fontsize=14)
plt.xlabel("City / Location", fontsize=12,)
plt.ylabel("Minimum Temperature (°C)", fontsize=12)
plt.xticks(rotation=45) 
plt.tight_layout()
plt.show()
```
