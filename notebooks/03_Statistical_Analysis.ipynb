### Trend Analysis

```python
# Monthly Temperature Trend
monthly_temp = df.groupby(['year','month'])['temperature_celsius'].mean().reset_index()

monthly_temp['date'] = pd.to_datetime(monthly_temp[['year','month']].assign(day=1))

plt.figure(figsize=(10,5))
plt.plot(monthly_temp['date'], monthly_temp['temperature_celsius'])
plt.title("Monthly Temperature Trend")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
yearly_temp = df.groupby('year')['temperature_celsius'].mean()
yearly_temp.pct_change()
```

```python
monthly_temp['rolling_mean'] = monthly_temp['temperature_celsius'].rolling(3).mean()

plt.plot(monthly_temp['date'], monthly_temp['rolling_mean'])
plt.title("Smoothed Monthly Temperature Trend")
plt.show()
```

```python
# Monthly Rainfall Trend
monthly_rain = df.groupby(['year','month'])['precip_mm'].sum().reset_index()

monthly_rain['date'] = pd.to_datetime(monthly_rain[['year','month']].assign(day=1))

plt.figure(figsize=(10,5))
plt.plot(monthly_rain['date'], monthly_rain['precip_mm'])
plt.title("Monthly Rainfall Trend")
plt.xlabel("Date")
plt.ylabel("Precipitation (mm)")
plt.show()
```

# Task 3
#### Compare weather events across region


#### Average Temperature by Country

```python
top_hot = df.groupby('country')['temperature_celsius'] \
            .mean() \
            .sort_values(ascending=False) \
            .head(10)

plt.figure(figsize=(8,5))
top_hot.sort_values().plot(kind='barh')
plt.title("Top 10 Hottest Countries (Average Temperature)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Country")
plt.show()
```

#### Rainfall Comparison

```python
top_rain = df.groupby('country')['precip_mm'] \
             .mean() \
             .sort_values(ascending=False) \
             .head(10)

plt.figure(figsize=(8,5))
top_rain.sort_values().plot(kind='barh')
plt.title("Top 10 Wettest Countries (Average Rainfall)")
plt.xlabel("Precipitation (mm)")
plt.ylabel("Country")
plt.show()
```

#### Wind Speed Comparison

```python
top_wind = df.groupby('country')['wind_kph'] \
             .mean() \
             .sort_values(ascending=False) \
             .head(10)

plt.figure(figsize=(8,5))
top_wind.sort_values().plot(kind='barh')
plt.title("Top 10 Windiest Countries")
plt.xlabel("Wind Speed (km/h)")
plt.ylabel("Country")
plt.show()
```

#### Humidity comparision

```python
top_humidity = df.groupby('country')['humidity'] \
                 .mean() \
                 .sort_values(ascending=False) \
                 .head(10)

plt.figure(figsize=(8,5))
top_humidity.sort_values().plot(kind='barh')
plt.title("Top 10 Most Humid Countries")
plt.xlabel("Humidity (%)")
plt.ylabel("Country")
plt.show()
```

#### Air Quality Comparison

```python
top_pollution = df.groupby('country')['air_quality_PM2.5'] \
                  .mean() \
                  .sort_values(ascending=False) \
                  .head(10)

plt.figure(figsize=(8,5))
top_pollution.sort_values().plot(kind='barh')
plt.title("Top 10 Most Polluted Countries (PM2.5)")
plt.xlabel("PM2.5 Level")
plt.ylabel("Country")
plt.show()
```

#### Climate Variability  (this shows unstable climates)

```python
top_variability = df.groupby('country')['temperature_celsius'] \
                    .std() \
                    .sort_values(ascending=False) \
                    .head(10)

plt.figure(figsize=(8,5))
top_variability.sort_values().plot(kind='barh')
plt.title("Top 10 Countries with Highest Temperature Variability")
plt.xlabel("Temperature Std Dev")
plt.ylabel("Country")
plt.show()
```

```python
import plotly.express as px

country_summary = df.groupby('country').agg({
    'temperature_celsius':'mean',
    'precip_mm':'mean',
    'wind_kph':'mean',
    'humidity':'mean'
}).reset_index()
```

```python
fig = px.scatter(
    country_summary,
    x='temperature_celsius',
    y='precip_mm',
    size='wind_kph',
    color='humidity',
    hover_name='country',
    title='Global Weather Comparison Across Countries',
    labels={
        'temperature_celsius':'Avg Temperature (°C)',
        'precip_mm':'Avg Rainfall (mm)',
        'wind_kph':'Avg Wind Speed (km/h)',
        'humidity':'Avg Humidity (%)'
    }
)

fig.show()
```

```python
season_country_summary = df.groupby(['season','country']).agg({
    'temperature_celsius':'mean',
    'precip_mm':'mean',
    'wind_kph':'mean',
    'humidity':'mean'
}).reset_index()

season_country_summary.head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for season in df['season'].unique():
    seasonal_data = season_country_summary[season_country_summary['season'] == season]
    top_temp = seasonal_data.sort_values(by='temperature_celsius', ascending=False).head(10)

    plt.figure(figsize=(8,5))
    sns.barplot(x='temperature_celsius', y='country', data=top_temp)
    plt.title(f"Top 10 Hottest Countries in {season}")
    plt.xlabel("Avg Temperature (°C)")
    plt.ylabel("Country")
    plt.show()
```

```python
season_temp_pivot = season_country_summary.pivot(
    index='country',
    columns='season',
    values='temperature_celsius'
)

season_temp_pivot.head()
```

```python

fig = px.imshow(
    season_temp_pivot,
    labels=dict(x="Season", y="Country", color="Avg Temperature (°C)"),
    title="Seasonal Temperature Comparison Across Countries",
    aspect="auto"
)

fig.show()
```
