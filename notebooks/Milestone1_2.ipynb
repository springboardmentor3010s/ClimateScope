# Milestone 1 – Data Preparation & Initial Analysis
This notebook performs data loading, inspection, cleaning, anomaly removal, feature engineering, aggregation, and saving processed datasets.

# mile stone 1

```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_csv('/Users/garikapatiaishwarya/Desktop/climatescope/data/raw/GlobalWeatherRepository.csv')
df.head()
```

```python
# Inspect structure
df.shape, df.columns, df.info(), df.describe()
```

```python
# Convert datetime
df['last_updated'] = pd.to_datetime(df['last_updated'])
df['last_updated'].head()
```

```python
# Missing values summary
missing_counts = df.isnull().sum()
missing_percent = (missing_counts / len(df)) * 100
missing_summary = pd.DataFrame({'missing_count': missing_counts, 'missing_percent': missing_percent})
missing_summary.sort_values(by='missing_percent', ascending=False)
```

```python
# Handle missing values (drop critical nulls)
df_cleaned_nulls = df.dropna(subset=[
    'temperature_celsius','humidity','precip_mm','wind_kph','uv_index','last_updated','country'
])
df_cleaned_nulls.shape
```

```python
# Remove unrealistic anomalies
df_cleaned_nulls = df_cleaned_nulls[
    (df_cleaned_nulls['temperature_celsius'].between(-50, 60)) &
    (df_cleaned_nulls['precip_mm'] >= 0) &
    (df_cleaned_nulls['wind_kph'] <= 200)
]
df_cleaned_nulls.shape
```

```python
# Standardize country names
df_cleaned_nulls['country'] = df_cleaned_nulls['country'].str.strip().str.title()
```

```python
# Fahrenhit to celsius: 

df["temperature_celsius"] = (df["temperature_fahrenheit"] - 32) * 5/9

# mph → km/h

df["humidity_norm"] = (df["humidity"] - df["humidity"].min()) / \
                      (df["humidity"].max() - df["humidity"].min())

```

```python
# Select relevant columns
df_clean = df_cleaned_nulls[[
    'country','location_name','latitude','longitude','last_updated',
    'temperature_celsius','humidity','precip_mm','wind_kph','condition_text','uv_index'
]]
```

```python
# Feature engineering
df_clean['year'] = df_clean['last_updated'].dt.year
df_clean['month'] = df_clean['last_updated'].dt.month
df_clean['month_name'] = df_clean['last_updated'].dt.month_name()
```

```python
# Data coverage analysis
date_range = (df_clean['last_updated'].min(), df_clean['last_updated'].max())
num_countries = df_clean['country'].nunique()
records_per_country = df_clean['country'].value_counts().head()
date_range, num_countries, records_per_country
```

```python
# Monthly aggregation
monthly_df = df_clean.groupby(['country','year','month'], as_index=False).agg({
    'temperature_celsius': 'mean',
    'humidity': 'mean',
    'precip_mm': 'sum',
    'wind_kph': 'mean',
    'uv_index': 'mean'
})
monthly_df.head()
```

```python
# Save cleaned daily dataset
df_clean.to_csv('/Users/garikapatiaishwarya/Desktop/climatescope/data/processed/global_weather_cleaned_daily.csv', index=False)

# Save aggregated monthly dataset
monthly_df.to_csv('/Users/garikapatiaishwarya/Desktop/climatescope/data/processed/global_weather_cleaned_monthly.csv', index=False)
```

# Milestone 2 – Core Analysis & Visualization Design

This notebook performs statistical analysis, correlation analysis, seasonal trend exploration, extreme weather event detection, regional climate comparison, visualization generation, and export of analytical summaries to support dashboard design and reporting.

# milestone 2

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Use datasets already created in Milestone 1
daily_df = df_clean.copy()        # cleaned daily data
monthly_df = monthly_df.copy()    # monthly aggregated data

```

```python

# --------------------------------
# 1. Statistical Distribution
# --------------------------------
print("Statistical Summary:")
print(daily_df.describe())

numeric_cols = ["temperature_celsius", "humidity", "wind_kph", "precip_mm"]

for col in numeric_cols:
    plt.figure()
    sns.histplot(daily_df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python

# --------------------------------
# 2. Correlation Analysis
# --------------------------------
plt.figure()
corr = daily_df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python

# ==========================================
# 3. Feature Distributions
# ==========================================
for col in numeric_cols:
    plt.figure()
    sns.histplot(daily_df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python

# ==========================================
# 4. Seasonal Trends (Temperature, Rainfall, Humidity)
# ==========================================
plt.figure()
sns.lineplot(data=monthly_df, x="month", y="temperature_celsius", hue="country")
plt.title("Monthly Temperature Trends by Country")
plt.tight_layout()
plt.show()

plt.figure()
sns.lineplot(data=monthly_df, x="month", y="precip_mm", hue="country")
plt.title("Monthly Rainfall Trends by Country")
plt.tight_layout()
plt.show()

plt.figure()
sns.lineplot(data=monthly_df, x="month", y="humidity", hue="country")
plt.title("Monthly Humidity Trends by Country")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 5. Extreme Weather Detection (Quantile Method)
# ==========================================
high_temp = daily_df["temperature_celsius"].quantile(0.99)
low_temp = daily_df["temperature_celsius"].quantile(0.01)

extreme_events = daily_df[
    (daily_df["temperature_celsius"] >= high_temp) |
    (daily_df["temperature_celsius"] <= low_temp)
]

print("\nTotal Extreme Temperature Events:", len(extreme_events))


# Scatter plot of extreme events
plt.figure()
sns.scatterplot(
    data=extreme_events,
    x="last_updated",
    y="temperature_celsius",
    hue="country"
)
plt.title("Extreme Temperature Events Over Time")
plt.tight_layout()
plt.show()
```

```python

# ==========================================
# 6. Extreme Events per Year
# ==========================================
daily_df["year"] = daily_df["last_updated"].dt.year

extreme_per_year = extreme_events.groupby("year").size().reset_index(name="extreme_count")

print("\nExtreme Events Per Year:")
print(extreme_per_year)

plt.figure()
sns.barplot(data=extreme_per_year, x="year", y="extreme_count")
plt.title("Extreme Temperature Events per Year")
plt.tight_layout()
plt.show()
```

```python

# ==========================================
# 7. Countries with Most Extreme Events
# ==========================================
extreme_by_country = extreme_events["country"].value_counts().reset_index()
extreme_by_country.columns = ["country", "extreme_count"]

print("\nTop Countries with Extreme Events:")
print(extreme_by_country.head())

plt.figure()
sns.barplot(data=extreme_by_country.head(10), x="extreme_count", y="country")
plt.title("Top 10 Countries with Most Extreme Temperature Events")
plt.tight_layout()
plt.show()
```

```python

# ==========================================
# 8. Regional Weather Comparison
# ==========================================
regional_stats = daily_df.groupby("country").agg({
    "temperature_celsius": "mean",
    "precip_mm": "sum",
    "humidity": "mean"
}).reset_index()

print("\nRegional Weather Statistics:")
print(regional_stats.head())


# Top hottest countries
plt.figure()
sns.barplot(
    data=regional_stats.sort_values("temperature_celsius", ascending=False).head(10),
    x="temperature_celsius",
    y="country"
)
plt.title("Top 10 Hottest Countries (Average Temperature)")
plt.tight_layout()
plt.show()


# Boxplot comparison across countries
plt.figure()
sns.boxplot(data=daily_df, x="country", y="temperature_celsius")
plt.xticks(rotation=90)
plt.title("Temperature Distribution Across Countries")
plt.tight_layout()
plt.show()

```

```python
# 9

# ---- Seasonal trends for rainfall & humidity ----
plt.figure()
sns.lineplot(data=monthly_df, x="month", y="precip_mm", hue="country")
plt.title("Monthly Rainfall Trends by Country")
plt.show()

plt.figure()
sns.lineplot(data=monthly_df, x="month", y="humidity", hue="country")
plt.title("Monthly Humidity Trends by Country")
plt.show()
```

```python
# 10

# ---- Extreme events per year ----
daily_df["year"] = pd.to_datetime(daily_df["last_updated"]).dt.year

extreme_per_year = extreme_events.groupby("year").size().reset_index(name="extreme_count")
print(extreme_per_year)

plt.figure()
sns.barplot(data=extreme_per_year, x="year", y="extreme_count")
plt.title("Extreme Temperature Events per Year")
plt.show()
```

```python
# 11

# ---- Countries with most extreme events ----
extreme_by_country = extreme_events["country"].value_counts().reset_index()
extreme_by_country.columns = ["country", "extreme_count"]

print(extreme_by_country.head())

plt.figure()
sns.barplot(data=extreme_by_country.head(10), x="extreme_count", y="country")
plt.title("Top 10 Countries with Most Extreme Temperature Events")
plt.show()
```

```python
# 12

# ---- Regional aggregated statistics ----
regional_stats = daily_df.groupby("country").agg({
    "temperature_celsius": "mean",
    "precip_mm": "sum",
    "humidity": "mean"
}).reset_index()

print(regional_stats.head())

plt.figure()
sns.barplot(
    data=regional_stats.sort_values("temperature_celsius", ascending=False).head(10),
    x="temperature_celsius",
    y="country"
)
plt.title("Top 10 Hottest Countries (Average Temperature)")
plt.show()
```
